```python
import os
import requests
import json
import time
import pandas as pd
import matplotlib.pyplot as plt

if not os.path.exists("data"):
    os.makedirs("data")
 

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36",
    "Accept-Encoding": "gzip, deflate, br",
    "Accept-Language": "en-US,en;q=0.9",
    "Referer": "https://www.nseindia.com/option-chain",
    "Connection": "keep-alive"
}

def fetch_option_chain_data(url, symbol):
    try:
        session = requests.Session()
        session.headers.update(HEADERS)
        session.get("https://www.nseindia.com", timeout=5)

        response = session.get(url, timeout=10)
        response.raise_for_status() 
        data = response.json()


        file_path = f"data/{symbol}_option_chain.json"
        with open(file_path, "w") as file:
            json.dump(data, file, indent=4)

        print(f"\n✅ Data for {symbol} fetched and saved at {file_path}")
        return data
    except requests.exceptions.RequestException as e:
        print(f"❌ Error fetching data for {symbol}: {e}")
        return None

def parse_option_chain_data(data, expiry_date):
    """Extract relevant columns, filter 5 rows above and below the current price."""
    option_data = data.get("records", {}).get("data", [])
    current_price = data.get("records", {}).get("underlyingValue", None)

    if current_price is None:
        print("⚠ Could not fetch current price.")
        return None

    rows = []
    for entry in option_data:
        strike_price = entry.get("strikePrice")
        expiry = entry.get("expiryDate")

        if expiry == expiry_date:
            call_data = entry.get("CE", {})
            put_data = entry.get("PE", {})

            rows.append({
                "Strike Price": strike_price,
                "Call OI": call_data.get("openInterest", 0),
                "Call LTP": call_data.get("lastPrice", 0),
                "Put OI": put_data.get("openInterest", 0),
                "Put LTP": put_data.get("lastPrice", 0),
            })

    df = pd.DataFrame(rows)


    df_sorted = df.sort_values("Strike Price").reset_index(drop=True)
    nearest_index = (df_sorted["Strike Price"] - current_price).abs().idxmin()


    start_idx = max(nearest_index - 5, 0)
    end_idx = min(nearest_index + 6, len(df_sorted))  
    df_filtered = df_sorted.iloc[start_idx:end_idx].reset_index(drop=True)

    print(f"\n🔹 {expiry_date} Expiry | Current Price: {current_price}")
    print(f"\n🔹 Showing 5 rows above and below the current price:\n")
    print(df_filtered.to_string(index=False)) 

    return df_filtered

def visualize_option_chain(df, symbol):
    """Plot Strike Price vs Open Interest for Call and Put options."""
    plt.figure(figsize=(12, 6))

    plt.bar(df["Strike Price"], df["Call OI"], color="blue", alpha=0.5, label="Call OI")
    plt.bar(df["Strike Price"], df["Put OI"], color="red", alpha=0.5, label="Put OI")

    plt.xlabel("Strike Price")
    plt.ylabel("Open Interest")
    plt.title(f"{symbol} Option Chain - Open Interest")
    plt.legend()
    plt.grid()
    plt.show()

if __name__ == "__main__":

    NIFTY_URL = "https://www.nseindia.com/api/option-chain-indices?symbol=NIFTY"
    HDFCBANK_URL = "https://www.nseindia.com/api/option-chain-equities?symbol=HDFCBANK"
    while True:
        for symbol, url in [("NIFTY", NIFTY_URL), ("HDFCBANK", HDFCBANK_URL)]:
            data = fetch_option_chain_data(url, symbol)
            if data:
                expiry_date = data["records"]["expiryDates"][0]  
                df = parse_option_chain_data(data, expiry_date)

                if df is not None:
                    visualize_option_chain(df, symbol)

        print("\n⏳ Sleeping for 3 minutes before fetching new data...\n")
        time.sleep(180) 
```
